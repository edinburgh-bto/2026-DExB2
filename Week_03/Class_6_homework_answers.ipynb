## Some categorical data

Lastly, we will generate some data with a categorical variable and numerical variables that depends on the categorical variable. The idea is a simulated dataset that is similar in style to the iris or penguins datasets that we have used elsewhere in the course.

Precisely, we have 3 species: major, big, and minor.

Each species has a different mean width.
The values are normally distributed around those widths.
The width is proportional to the width with some extra variability, but the proportion is different for different species.

To simulate this, we'll first define the population mean width and proportion height vs width. These are usually the parameters we'd want to discover in a statistical analysis.


```python
# set up mean values into a data frame indexed by species

means_by_species = pd.DataFrame(
    {"species" : ['big', 'major', 'minor'],
     "mean_width" : [8.1, 8.0, 6.0],
     "height_vs_width" : [1.4, 1.2, 1.1]
    })

means_by_species.index = means_by_species["species"]
means_by_species
```

We next check that we got the indexing right, so can access values using `df.loc`.

Feel free to do some checks yourself, or to re-check the `loc` documentation.

```python
means_by_species.loc["major", "mean_width"]
```

Now we use these values to simulate a random sample from the species.

```python
# reseed random number generator with a different seed
rng_seed20 = np.random.default_rng(seed=20)

# generate a data frame with 20 random samples of species
df_categorical20 = pd.DataFrame(
    {"species" : rng_seed20.choice(['major', 'big', 'minor'], size = 20)
    })

# simulate widths as variables with the given means
df_categorical20["width"] = rng_seed20.normal(
    loc = means_by_species.loc[df_categorical20["species"],"mean_width"],
    scale = 1.0)

# create a height_expected_from_width vector to reflect scaling by species
height_expected_from_width = (
    df_categorical20["width"].to_numpy() * 
    means_by_species.loc[df_categorical20["species"], "height_vs_width"].to_numpy()
)

# simulate a width that's normally distributed around height_expected_from_width 
df_categorical20["height"] = rng_seed20.normal(
   loc = height_expected_from_width,
   scale = 0.2 )

df_categorical20
```
