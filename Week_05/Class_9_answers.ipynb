Weddell seals are remarkable elite divers capable of hunting beneath the Antarctic ice shelf for extended periods.

Marine biologists were interested in the physiological limits of the seals by recording dive durations on 120 seal foraging trips.

The data collected by the researchers are in the file `../Datasets/weddell_seal_dives_full.csv`.

---

- Read in the dataset and print it out to see what it looks like.

````python
dives = pd.read_csv('../Datasets/weddell_seal_dives_full.csv')
dives
````

- Use an appropriate graph to visually examine the dive durations.

````python
sns.histplot(dives['dive_duration_min']);
````

- Calculate the sample size, sample mean and standard deviation. 

<div class="alert alert-info">

Answer

- n: 120 dives
- xbar: 46.0 mins
- s: 3.5 mins

</div>

````python
n = dives['dive_duration_min'].count()
xbar = dives['dive_duration_min'].mean()
s = dives['dive_duration_min'].std()

print(f'n: {n} dives')
print(f'xbar: {xbar:.1f} mins')
print(f's: {s:.1f} mins')
````

- Calculate the standard error in the mean.

<div class="alert alert-info">

Answer

SEM: 0.32 mins

</div>


````python
sem = s/np.sqrt(n)
print(f'SEM: {sem:.2f} mins')
````

- Calculate the 95% CI.

<div class="alert alert-info">

Answer

- 95% CI: 46.0 ± 0.6 mins
- 95% CI: [45.3, 46.6] mins

</div>


````python
tv = t.ppf(0.975, df=n-1)
u = tv*sem
print(f'95% CI: : {xbar:.1f} ± {u:.1f} mins')

ci_lower = xbar - u
ci_upper = xbar + u
print(f'95% CI: [{ci_lower:.3g}, {ci_upper:.3g}] mins')
````

<div class="alert alert-success">

Concepts to take away from this question.

1. A population parameter is some characteristic of a population. For example, the **mean** dive duration of Weddell seals. There are many other types of parameters, e.g., median, mode, standard deviation, correlation coefficient, and so on.
2. A sample (or descriptive) statistic is a characteristic of a sample drawn from that population.  For example, the **mean** dive duration of ten Weddell seals.
3. A sampling distribution of a sample statistic is the hypothetical distribution of that statistic (think of a histogram) if you could sample the population an infinite number of times.
4. Being a distribution, a sampling distribution has a standard deviation which measures its width. 
5. The technical term for the standard deviation of a sampling distribution is the standard error. For example, standard error of the mean of ten dive durations.
6. Standard errors are absolutely necessary when estimating a population parameter. They tell us how much uncertainty, or how precise, our estimate is of a population parameter.
7. The standard error of a mean is $\frac{s}{\sqrt{n}}$. So larger sample sizes have smaller standard errors, and less variable populations have smaller standard errors.
8. For small sample sizes, SEM underestimates the uncertainty in the estimate of the population mean. Instead we use the 95% confidence interval (which accounts for small sample sizes).
9. Always give either the standard error or the 95% confidence interval in a scientific report, paper or poster otherwise people reading it won't know how precise your estimate is.

# **Question 2:** Estimating the evolutionary change of ground finch beak depths on the Galápagos islands

This example is from Biology 1A.

Peter and Rosemary Grant and their colleagues have studied Galápagos finch species every year since 1976. One such species is the medium ground finch on the Galápagos island of Daphne Major.
<div>
<img src="Images/ground_finch.jpg" width='50%' title="Judy Gallagher CC BY 2.0"/>
</div>

The medium ground finch feeds on seeds. The finches have inherited variation in beak shape with some individuals having wide, deep beaks and others having thinner, shallower beaks. Large-beaked birds feed more efficiently on large, hard seeds, whereas smaller-beaked birds feed more efficiently on small, soft seeds. 

During 1977 a drought altered vegetation on the island. After the drought the number of seeds declined dramatically. The decline was greater in small, soft seeds than in large, hard seeds. Therefore, the Grants and colleagues hypothesised that large-beaked birds were better able to survive the drought than the small-beaked birds. 

As beak size is inherited from parent to offspring the theory of natural selection predicts that the drought would cause selection for finches with larger beaks. We can test this hypothesis by comparing average beak sizes of medium ground finches on Daphne Major before (1976) and after (1978) the drought.

The beak depths (in mm) collected by the Grants are in the file `../Datasets/finches beak depth.csv`.

---

- Read in the dataset and print it out to see what it looks like.

````python
beaks = pd.read_csv('../Datasets/finches beak depth.csv')
beaks
````

- Use an appropriate graph to visually compare beak depths between 1976 and 1978.

````python
# A swarm plot combined with a box plot to show the distribution of beak depths across years is a good choice. The swarm plot will show individual data points, while the box plot will summarize the distribution with quartiles and outliers.

sns.swarmplot(data=beaks, x='year', y='depth')
sns.boxplot(data=beaks, x='year', y='depth', color='lightblue');
````

- For the 1976 beak depths only, calculate their sample mean ($\bar{x}_{1976}$), their standard deviation ($s_{1976}$), their SEM ($\text{SEM}_{1976}$) and their 95% CI.

<div class="alert alert-success">

Hint: you will first need to subset just the 1976 beak depths from the dataframe. There are two ways of doing this:

<div class="alert alert-info">

Answer

- n_1976:      75 beaks
- xbar_1976:   9.49 mm
- s_1976:      0.94 mm
- SEM_1976:    0.11 mm
- 95% CI_1976: [9.27, 9.71] mm

</div>


````python
beaks_1976 = beaks.query('year == 1976')

n_1976 = beaks_1976['depth'].count()
xbar_1976 = beaks_1976['depth'].mean()
s_1976 = beaks_1976['depth'].std()
sem_1976 = beaks_1976['depth'].sem()
tv_1976 = t.ppf(0.975, df=n_1976-1)
ci_low_1976 = xbar_1976 - tv_1976*sem_1976
ci_high_1976 = xbar_1976 + tv_1976*sem_1976

print(f'n_1976:      {n_1976} beaks')
print(f'xbar_1976:   {xbar_1976:.2f} mm')
print(f's_1976:      {s_1976:.2f} mm')
print(f'SEM_1976:    {sem_1976:.2f} mm')
print(f'95% CI_1976: [{ci_low_1976:.2f}, {ci_high_1976:.2f}] mm')
````

- Now do the same for the 1978 beak depths.

<div class="alert alert-info">

Answer

- n_1978:      60 beaks
- xbar_1978:   9.98 mm
- s_1978:      0.84 mm
- SEM_1978:    0.11 mm
- 95% CI_1978: [9.77, 10.20] mm

</div>


````python
beaks_1978 = beaks.query('year == 1978')

n_1978 = beaks_1978['depth'].count()
xbar_1978 = beaks_1978['depth'].mean()
s_1978 = beaks_1978['depth'].std()
sem_1978 = beaks_1978['depth'].sem()
tv_1978 = t.ppf(0.975, df=n_1978-1)
ci_low_1978 = xbar_1978 - tv_1978*sem_1978
ci_high_1978 = xbar_1978 + tv_1978*sem_1978

print(f'n_1978:      {n_1978} beaks')
print(f'xbar_1978:   {xbar_1978:.2f} mm')
print(f's_1978:      {s_1978:.2f} mm')
print(f'SEM_1978:    {sem_1978:.2f} mm')
print(f'95% CI_1978: [{ci_low_1978:.2f}, {ci_high_1978:.2f}] mm')
````

Next we want to estimate the **difference** in beak depths between 1976 and 1978.

The difference in the means, $d$, is simply 

$$d=\bar{x}_{1978}-\bar{x}_{1976}$$

However, because there is uncertainty in the two means it stands to reason there must also be uncertainty in their difference.

The uncertainty in the difference is called the **standard error of the difference in the means**. The formula for this is:

$$\text{standard error of the difference in the means} = \sqrt{\text{SEM}^2_{1978}+\text{SEM}^2_{1976}}$$

In words, you square the standard errors of the means for 1976 and 1978, add them and take the square root.

---

- Calculate the difference in the sample means and the standard error of the difference in the means. 

<div class="alert alert-info">

Answer

- Mean difference: 0.49 mm
- SE of difference: 0.15 mm

</div>


````python
d = xbar_1978 - xbar_1976
print(f'Mean difference: {d:.3f} mm')
sed = np.sqrt(sem_1978**2 + sem_1976**2)
print(f'SE of difference: {sed:.3f} mm')
````

Now we want to calculate the 95% CI of the difference in the means. We use the same formulae but modified for the difference in the means.

$$\text{low bound} = d - t_{n_{1976}+n_{1978}-2} \times \text{SE of difference}$$
$$\text{high bound} = d + t_{n_{1976}+n_{1978}-2} \times \text{SE of difference}$$

Note that to find *t* the degrees of freedom are the sum of the two sample sizes minus 2.

---

- Calculate the 95% CI of the difference in the means.

<div class="alert alert-info">

Answer

- 95% CI of difference: 0.49 ± 0.30 mm
- 95% CI of difference: [0.19, 0.80] mm
</div>



````python
tv = t.ppf(0.975, df=n_1976+n_1978-2)
u = tv*sed
ci_low = d - u
ci_high = d + u

print(f'95% CI of difference: {d:.2f} ± {u:.2f} mm')
print(f'95% CI of difference: [{ci_low:.2f}, {ci_high:.2f}] mm')
````

# **Question 3:** What proportion of hospital patients have vancomycin-resistant intestinal *Enterococcus* bacteria?

*Entorococcus* is a genus of bacteria that can be found in the gastrointestinal tracts of humans and animals. Some species of *Enterococcus* are known to be opportunistic pathogens, meaning they can cause infections in individuals with weakened immune systems or when they enter parts of the body where they are not normally found. These infections can include urinary tract infections, bloodstream infections, and wound infections. 

Some strains of *Entorococcus* are also known to be resistant to multiple antibiotics including vancomycin, which can make infections difficult to treat. The presence of *Enterococcus* in clinical settings is often a concern due to its potential to cause healthcare-associated infections and its ability to spread antibiotic resistance genes.

Researchers want to know what proportion of hospital patients in general have vancomycin-resistant intestinal *Enterococcus* bacteria. To answer this question they isolated *Enterococcus* from the faeces of 100 randomly sampled hospital patients and tested them for resistance to vancomycin. The data are given in the file `../Datasets/antibiotic_resistance.csv`.

Each row in the dataset is a single patient and whether they have vancomycin-resistant or non-resistance *Enterococcus* bacteria.

---

9% is, of course, an estimate because only 100 people were sampled out of the many millions that visit hospital. 

We therefore need to calculate the standard error in that estimate. 

You are going to use two methods to do this:
1. Using the known formula for the standard error of a proportion, and
2. using Monte-Carlo simulation of the sampling distribution pf a proportion.

If you have done these correctly you will get the same answer for methods 1) and 2).

### Method 1: Formula

The formula for the standard error of a proportion is

$$\text{SE}_{\hat{p}} = \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$$

where $\hat{p}$ is the estimated proportion (e.g., 0.09) and $n$ is the sample size (e.g., 100).

---

- Calculate the standard error of the proportion using Method 1.

<div class="alert alert-info">

Answer to 3dp

standard error of phat = 0.029
</div>

````python
SE_phat = np.sqrt(phat*(1-phat)/n)
SE_phat
````

Next we should find the 95% confidence interval of the proportion. Unlike for the confidence interval of a mean, there is no exact formula for the confidence interval of a proportion. Instead there is an approximate formula. We first have to calculate $p^\prime$ using this formula:

$$p^\prime = \frac{X+2}{n+4}$$

where $X$ is the number of patients with resistant bacteria (i.e., 9).

Then calculate $\text{SE}_{p^\prime}$ like so:

$$\text{SE}_{p^\prime} = \sqrt{\frac{p^\prime(1-p^\prime)}{n+4}}$$

Then the low and high bounds of the 95% CI are

- low bound = $p^\prime - t_{n-1}\times\text{SE}_{p^\prime}$
- high bound = $p^\prime + t_{n-1}\times\text{SE}_{p^\prime}$

---

- Calculate the 95% CI of the proportion of people with resistant bacteria.

<div class="alert alert-info">

Answer to 3dp

95% CI = [0.046, 0.166]
</div>

````python
X = 9
n = 100
tv = t.ppf(0.975, df=n-1)
pp = (X+2)/(n+4)
SE_pp = np.sqrt(pp*(1-pp)/(n+4))
u = tv*SE_pp
low = pp - u
high = pp + u
print(f'95% CI = [{low:.3f}, {high:.3f}]')
````

### Method 2: Monte-Carlo simulation

Our aim is to simulate thousands of random proportions to construct a sampling distribution of proportions. First we need to understand what we want to simulate.

In the dataset of `antibiotic_resistance.csv` we have 100 people of which 9 have resistant bacteria and 91 do not. In other words, 9% of people have resistant bacteria. Or put another way, if we were to randomly select an individual person from these 100, there is a 9 in 100 chance (a probability of 0.09) of choosing one with resistant bacteria.

Now we perform a thought experiment. Let's assume we have a very large hypothetical population of people of which exactly 9% have resistant bacteria and 91% don't. If we were to randomly sample 100 of these people how many would have resistant bacteria? 

The answer, of course, depends on who we sample. Sampling is random (like rolling a dice) so we may actually sample no people with resistant bacteria, or we may sample 1 person with resistant bacteria, or 2 people, or 3, and so on all the way up to 100 people (although that would be highly unlikely).

If we were to take thousands of samples of 100 people, on average we would expect 9 of them to have resistant bacteria. But in any particular sample, the number of people with resistant bacteria could be any number from 0 to 100. 

To simulate sampling a random number of people with resistant bacteria we use a **Binomial distribution**. The python code to generate a random number of people with resistant bacteria in 100 randomly sampled people given that 9% of them have resistant bacteria is simply

```python
m = binomial(n=100, p=0.09)

print(f'{m} people have resistant bacteria out of 100 randomly sampled people')
```

- Try this in the code cell below. Each time you run the code cell you will get a different number of people with resistant bacteria.

````python
m = binomial(n=100, p=0.09)
print(f'{m} people have resistant bacteria out of 100 randomly sampled people')
````

- Convert the number of people with resistant bacteria to a proportion by dividing by the number of people sampled

````python
phat = m / 100
print(f'{phat:.0%} of people have resistant bacteria out of 100 randomly sampled people')
````

We want to sample 100 people from this hypothetical population thousands of times to construct the sampling distribution of the proportion of people with resistant bacteria. We can do this by passing `size=100000` to `binomial()` like so

```python
ms = binomial(n=100, p=0.09, size=100000)
```

````python
ms = binomial(n=100, p=0.09, size=100000)
print(ms)
````

- Now divide this list by the number of people sampled to obtain a list of 100,000 proportions.

````python
phats = ms / 100
print(phats)
````

To find the 95% CI using Monte-Carlo simulation you need to find the 2.5 and 97.5 percentiles of your simulated sampling distribution. 

What is a percentile? The *x*<sup>th</sup> percentile of some values is the value at which *x* percent of the values lie below that value and 100-*x* percent of values lie above that value. 

That's a bit of a mouthful and difficult to understand. Let's look at some made-up values to illustrate what we mean.

For example, in panel (A) of the figure below, some made-up values (nothing to do with proportions) have been split into quarters. A quarter of the data lie in each coloured region. The 50<sup>th</sup> percentile (also known as the median) is 10.0: half the values lie below 10.0 and half lie above 10.0. The 25<sup>th</sup> percentile (also called the first quartile) is 9.3. A quarter of the values lie below 9.3 and three-quarters lie above 9.3.

<div>

</div>

For the same values shown in panel (B) the 2.5<sup>th</sup> percentile is 8.0; 2.5% of values lie below 8.0 and 97.5% lie above it. The 97.5<sup>th</sup> percentile is 12.0; 97.5% of values lie below 12.0 and 2.5% lie above 12.0.

This means that 95% of the values lie between 8.0 and 12.0. 

Applying the same logic to the sampling distribution of proportions we can the find the 2.5<sup>th</sup> and 97.5<sup>th</sup> percentiles of the simulated proportions to give the 95% CI on the proportion of people with resistant bacteria.

The code to find the *x*<sup>th</sup> percentile of data in a list called `data` is

```python
v = np.percentile(data, x)
```

Calculating the standard error of a proportion is obviously far simpler with method 1 than with method 2. For most cases you will come across there is a formula for calculating a standard error (see Table 1).

*Table 1: Formulae for the standard error of common estimates.*
Estimate | Population parameter | Sample statistic | SE forumla
:-- | :-: | :-: | :-:
Mean | $\mu$ | $\bar{x}$ | $\frac{s}{\sqrt{n}}$
Difference in means | $\mu_1-\mu_2$ | $\bar{x}_1 - \bar{x}_2$ | $\sqrt{\text{SEM}^2_1+\text{SEM}^2_2}$
Proportion | $p$ | $\hat{p}$ | $\sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$

Google will help if you ever need the standard error of a less common estimate.

3. Many common descriptive statistics (mean, proportion, correlation, etc.) have a formula for the standard error.
4. But many descriptive statistics don't have a formula. For those we can use Monte-Carlo simulation (and other techniques) to simulate a sampling distribution from which a standard error can be found.
