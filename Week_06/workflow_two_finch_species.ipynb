### Imports

````python
import pandas as pd
import seaborn as sns
from statsmodels.formula.api import ols
````

### Load and print the data

First we read in the dataset and print it to see how it looks and what the variable names, types and values are.

````python
beaks = pd.read_csv('../Datasets/species_beak_depths.csv')
beaks
````

Variable names are `year`, `depth` and `species`.

- The variable `year` is categorical with two levels `1976` and `1978` representing the two years in which beak depth measurements were taken. Although year is numerical it is treated as categorical for this analysis. 
- The variable `depth` is a continuous numerical variable with units of mm.
- The variable `species` is categorical with two levels `Cactus finch` and `Ground finch`.

Next we need to identify the explanatory (independent) and response (dependent) variables. In this case it is very simple. `year` and `species` are the explanatory variables and `depth` is the response variable. 

### Explicitly set year as categorical

We are treating `year` as a categorical variable with two levels. But pandas and seaborn don't know that; to them it looks like a numerical value. This doesn't have much impact, but it does sometimes affect how things are plotted. So we will explicitly tell pandas that `year` is categorical using the following code.


````python
beaks['year'] = beaks['year'].astype('category')
````

### Plot the data

Now lets plot the data in a seaborn box plot. A swarmplot could also be used but it looks a lot messier and harder to see differences clearly.

One of the explanatory variable goes on the horizontal *x*-axis, the other is specified using the `hue` argument. The response variable `depth` goes on the vertical *y*-axis.

Here is a plot with `year` in the *x*-axis.

````python
sns.boxplot(data=beaks, x='year', y='depth', hue='species');
````

On first inspection it looks as if Cactus finch beak depths are on average about 1 mm larger then Ground finches and that beak depths increased for both species in 1978.

And here is a plot with `species` on the *x*-axis which shows that beak depths appear to have increased by half a millimetre in both species from 1976 to 1978.

````python
sns.boxplot(data=beaks, hue='year', y='depth', x='species');
````

In a report, thesis or poster you would never plot the same data twice like this. Once will do. You have to chose the best way to present your data in a single graph.

But when doing exploratory data analysis, as you're doing now, plotting the data in multiple ways is a good idea to get a feel for the data and explore the best ways to present it clearly.

### Model the data

Our model of the data is 

> beak depth **is related to** year *and* species

The way to code this model in python is

```python
formula = 'depth ~ C(year) + species'
```

The plus sign in mathematical formulae means "add two numbers together". Here is it means "and". So the formula reads "beak depth is related to year AND species. Also, because `species` has two levels which are words (i.e., "Cactus finch" and "Ground finch") python automatically knows that `species` is a categorical variable. This is unlike `year` which we explicitly had to tell python is a categorical variable using the form `C(year)`.

### Code and fit the model

Exactly as before, we code the model, fit it to the data and output a summary of the fit.

````python
# the model formula is coded as a string
formula = 'depth ~ C(year) + species'

# fit the model to the data
fit = ols(formula, beaks).fit()

# output summary table of the model fit
fit.summary().tables[1]
````

We'll also print the *P*-values as all are less than 0.001.

````python
fit.pvalues
````

### Reference levels

There are two categorical variables and each has a reference level. Python automatically choses levels which come first numerically or alphabetically. So, in this case the reference levels are `1976` for `year` and `Cactus finch` for `species`.


### Interpret the table

The value of `Intercept` is 10.5 mm, which is the average beak depth of Cactus finches in 1976 (the reference levels). 

The value of `C(year)[T.1978]` is 0.49 mm, which is the mean difference between 1976 and 1978 across both finch species. The is statistically significant with a *P* = 1.8 x 10<sup>-5</sup>, so highly incompatible with the null of no difference. Its 95% CI is [0.27, 0.70] mm. **In English, beak depths increased on average by about half a millimetre from 1976 to 1978.**

The value of `species[T.Ground finch]` is -1.01 mm, which is the mean difference between Cactus finches and Ground finches in both years. This is statistically significant with a *P* = 1.4 x 10<sup>-17</sup> and a 95% CI of [-1.23, -0.80] mm. So highly incompatible with null of no difference. **In English, Cactus finch beak depths are about one millimetre longer than Ground finch beak depths in both years.**

### Did the increase in beak depths differ between species?

This analysis has assumed that beak depths increased equally for both species from 1976 to 1978. In other words beak depths increased by half a millimetre for Ground finches and half a millimetre for Cactus finches. 

But that might not be the case.

We need to examine if the increase in beak depths differed between species. To do this we need to add another term to our model formula:

```python
formula = 'depth ~ C(year) + species + C(year):species'
```

In linear modelling terminology, the additional term `C(year):species` is called an interaction effect. The terms `C(year)` and `species` are called main effects.

When we fit this model to the data, the size of the interaction term tells us the **difference** in the increase in beak depths between the two species. If this difference is zero, then the increase in beak depths was the same for both species (i.e., 0.49 mm). But if this difference is significantly different from zero, then there is evidence that the increase in beak depths was larger for one species than the other.

Let's fit this model.

````python
formula2 = 'depth ~ C(year) + species + C(year):species'
fit2 = ols(formula2, beaks).fit()
fit2.summary().tables[1]
````

The value of the interaction term `C(year)[T.1978]:species[T.Ground finch]` is 0.02 mm. But the 95% CI contains zero ([-0.42, 0.46] mm) and *P* = 0.94, so a non-significant difference. The data are, therefore, compatible with the null hypothesis that the increase in beak depths was the same for both species.

If you compare this table with the previous one, you'll notice that all of the values are different, including the *P* values and the 95% CIs. So which table should we use to report the final conclusions? 

Well, because the interaction term in the second model was non-significant, it is not telling us anything new about the data. We should therefore discard this model and use the first, simpler model for our conclusions.

This principle is called Occam's razor. Among competing hypotheses that fit the data equally well, the one with the fewest assumptions should be selected. This principle prioritises simplicity in explanations or models.
