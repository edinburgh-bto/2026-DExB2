# Idea 1: Summary statistics and functions

We are going to remind ourselves what a mean is, and how to calculate one in Python.

Written as an equation: $\text{mean}(x) = \frac{1}{n} \sum^{n}_{i = 1} x_i$

Coded in Python: `mean(x) = sum(x) / len(x)`


```python
data1 = [1, 1, 3, 3, 6, 10]
```

```python
sum(data1)
```

```python
len(data1)
```

```python
sum(data1) / len(data1)
```

```python
mean(data1)
```

The `mean` function doesn't exist in *base* Python.

Let's define a function to calculate the mean. The idea of defining functions was covered in Biology 1C: Discovery.

Very nice explanation on functions from the Software Carpentry "Novice Python" lesson: https://swcarpentry.github.io/python-novice-inflammation/instructor/08-func.html

```python
def mymean(x):
    return sum(x) / len(x)
```

# Idea 2: Summary statistics of groups

We want to draw together 4 ideas that are related:

- **filtering data** and calculating a summary statistic)
- **grouping data** then calculating statistics for all groups
- **plotting filtered data** and visualizing summary stats
- **plotting data in groups**


Import the surveys dataset from the Datasets folder

```python
# import surveys
surveys = pd.read_csv("../Datasets/surveys.csv")
```

```python
surveys
```

Let's filter by species_id NL and calculate the mean weight using df.mean()

```python
surveys_NLonly = surveys[surveys["species_id"] == "NL"]
surveys_NLonly
```

```python
surveys_NLonly["weight"].mean()
```

Alternatively, you could group by species_id, then calculate the mean for every group.


```python
surveys_meanbyspecies = surveys.groupby("species_id").mean(numeric_only = True)

surveys_meanbyspecies
```

```python
surveys_meanbyspecies.loc["NL", "weight"]
```

So filtering then calculating summary statistics allows you to see (e.g. the mean) for just one group, the one you filtered by.

It's more efficient to use groupby, which can calculate the summary statistics for everything at once.
